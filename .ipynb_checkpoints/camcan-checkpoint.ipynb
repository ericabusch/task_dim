```python
def isc_kernel(data, sl_mask, myrad, bcvar):
    num_voxels_in_sl = sl_mask.shape[0] * sl_mask.shape[1] * sl_mask.shape[2]
    test_data = np.nan_to_num(data[0].ravel()) # flatten into  timepointsxvoxels
    train_data = np.nan_to_num(data[1].ravel()) # flatten into timepointsxvoxels
    # check for unique values 
    n1 = np.linalg.norm(train_data)
    n2 = np.linalg.norm(test_data)
    
    if n1 == 0 or n2 == 0: 
        r = np.nan 
    else: 
        r = np.corrcoef(train_data, test_data)[0,1]  # flatten both of these and correlate
    print(r, num_voxels_in_sl)
    return r
```

```python
def load_data(sub_id, task, subject_filter):
    
    # Load bold data and some header information so that we can save searchlight results as nifti later.
    nii = utils.get_subject_data(sub_id, task)
    # Load mask
    brain_mask = utils.get_intersect_mask(subject_filter=subject_filter)
    masker_wb = NiftiMasker(mask_img=brain_mask, standardize=True)
    masked_normed = masker_wb.fit_transform(nii)
    masked_nii = masker_wb.inverse_transform(masked_normed)
    
    bold_data = masked_nii.get_fdata()
    affine_mat = masked_nii.affine
    dimensions = masked_nii.header.get_zooms() 
    
    return bold_data, brain_mask.get_fdata(), affine_mat, dimensions
```

```python
def von_neumann_entropy(eigs: np.array, eps: float = 1e-3):
    eigenvalues = eigs.copy()

    eigenvalues = np.array(sorted(eigenvalues)[::-1])

    # Shift the negative eigenvalue(s) that occurred due to rounding errors.
    if eigenvalues.min() < 0:
        eigenvalues -= eigenvalues.min()

    # Drop the trivial eigenvalue corresponding to the indicator eigenvector.
    eigenvalues = eigenvalues[1:]

    # Drop the close-to-zero eigenvalue(s).
    eigenvalues = eigenvalues[eigenvalues >= eps]

    prob = eigenvalues / eigenvalues.sum()
    prob = prob + np.finfo(float).eps

    return -np.sum(prob * np.log(prob))
```

```python
import statsmodels.api as sm

def generate_autocorr_view(X, smooth_window=1):
    n_samples, n_features = X.shape
    # calculate and store the AC functions for each feature separately
    A_feat = np.empty((n_samples,n_features))
    for f in range(n_features):
        A_feat[:,f] = sm.tsa.acf(np.nan_to_num(X[:,f]), fft=False, nlags=n_samples-1, missing='conservative')
    A_feat = np.nanmean(A_feat, axis=1) # average over features to get one function
    acf = np.convolve(A_feat, np.ones(smooth_window), 'same') / smooth_window # rolling average
    if np.sum(acf != acf) > 0:
        dropoff = 0
    else:
        dropoff = np.where(acf  < 0)[0][0] # timepoint where rolling average drops off
    dropoff = dropoff
    print(dropoff)
    # Spread out the autocorr function
    M = np.zeros((n_samples, n_samples))
    for i in range(n_samples):
        for j in range(n_samples):
            if 0 < abs(i-j) < dropoff:
                M[i,j]=acf[abs(i-j)]
                M[j,i]=acf[abs(i-j)]
    diag = np.diag_indices(M.shape[0])
    M[diag] = 1
    return M

def row_normalize(M):
    # row normalize to turn to probabilities
    for row in M:
        if np.sum(row) == 0: # this should never be true
            continue
        row[:] /= np.sum(row)
    temp = M[0, 1]
    M[0, 1] = M[1, 0]
    M[1, 0] = temp
    return M 
```
