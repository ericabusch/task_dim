```python
all_dss = []
for subject in ['AVG']:# config.SUBJECTS[1:]:
    lhs, rhs, dss = [], [], []
    for sl_size in [20]:
        for run in range(1,5):
            try:
                l = np.load(f'results/sub-{subject}_ses-localizer_run-0{run}_rad{sl_size}_dimensionality_lh.npy')
                r = np.load(f'results/sub-{subject}_ses-localizer_run-0{run}_rad{sl_size}_dimensionality_rh.npy')
                lhs.append(l)
                rhs.append(r)
                ds = np.concatenate((l, r),axis=0)
                dss.append(ds)
            except:
                continue

        fig,ax=plt.subplots(2,2,figsize=(16,8))
        ax=ax.ravel()
        
        MINN = np.percentile(np.array(dss),5)
        MAXX = np.percentile(np.array(dss),95)
        
        
        
        all_dss.append(np.array(dss))
        for a, ds, r in zip(ax,dss, np.arange(1,5)):
            img, scale = bpl.brain_plot(ds, vmax=MAXX, vmin=MINN, cmap='magma',return_scale=True)
            a.imshow(img)
            a.axis('off')
            a.set_title(f'Run {r}')
        corr = np.mean(1-pdist(np.nan_to_num(all_dss[-1]),'correlation'))
        fig.suptitle(f'Subject {subject} localizer, sl_rad = {sl_size}, avg_corr={np.round(corr,4)}')
        fig.subplots_adjust(right=0.90)
        cbar_ax = fig.add_axes([0.92, 0.2, 0.02, 0.6])
        fig.colorbar(scale, cax=cbar_ax)
        
        plt.savefig(f'plots/sub-{subject}_ses-localizer_dim_rad{sl_size}.png')
        print(f'plots/sub-{subject}_ses-localizer_dim_rad{sl_size}.png')
        plt.clf()
        plt.close()
```

```python
# load in some data
ds = np.load(f'preprocessed_data/sub-01_ses-movie_task-movie_run-1_fsaverage5_cleaned_lh.npy')
#searchlight_data=ds[:, searchlight]
tphate_op = tphate.TPHATE(verbose=0, n_landmark=len(ds)).fit(ds)
dropoff = tphate_op.dropoff
dim = tphate_op._find_optimal_t() 
emd = tphate_op.transform(ds)
```

```python
from scipy.spatial.distance import pdist,squareform
corrmat = squareform(pdist(ds, 'correlation'))
print(corrmat.shape)
```

```python
plt.imshow(corrmat)
```

```python
tmax=100
ts=np.arange(tmax)
tph_entropy = tphate.vne.compute_von_neumann_entropy(tphate_op.diff_op, tmax)
ph_entropy = tphate.vne.compute_von_neumann_entropy(tphate_op.phate_diffop, tmax)
cdist_entropy = tphate.vne.compute_von_neumann_entropy(corrmat, tmax)
plt.plot(tph_entropy, label='tphate', c='r')
plt.plot(ph_entropy, label='phate', c='b')
#plt.plot(cdist_entropy, label='corr', c='g')

t_opt_tph = tphate.vne.find_knee_point(y=tph_entropy, x=ts)
t_opt_ph = tphate.vne.find_knee_point(y=ph_entropy, x=ts)
t_opt_c = tphate.vne.find_knee_point(y=cdist_entropy, x=ts)

plt.scatter(x=t_opt_tph, y=tph_entropy[t_opt_tph], c='r')
plt.scatter(x=t_opt_ph, y=ph_entropy[t_opt_ph], c='b')
#plt.scatter(x=t_opt_c, y=cdist_entropy[t_opt_c], c='g')

print(tphate_op.optimal_t, t_opt_tph, t_opt_ph, t_opt_c)
```

```python
for sub in config.SUBJECTS:
    ses = 'localizer'
    for inn, title in zip(['sl_res','sl_pca_res','sl_full_res'], ['tphate','pca','voxel']):
        rh = np.load(f'results/sub-{sub}_ses-{ses}_{inn}_rh.npy', allow_pickle=True)
        lh = np.load(f'results/sub-{sub}_ses-{ses}_{inn}_lh.npy', allow_pickle=True)
        ds = np.concatenate((lh,rh),axis=0)
        if title == 'tphate':
            labels = ['dropoff','dimension','clf_acc','clf_p','spatial_corr']
            cols = [0, 1, 2, 4]
        else:
            labels = ['dimension','clf_acc','clf_p','spatial_corr']
            cols = [0, 1, 3]
        for col, label in zip(cols, labels):
            fig,ax=plt.subplots()
            values_rh = rh[:, col]
            values_lh = lh[:, col]
            v = np.concatenate((values_lh, values_rh), axis=0)
            try:
                vmin,vmax=np.percentile(np.nan_to_num(v), [1,99])
            except:
                continue
            img, scale = bpl.brain_plot(v, vmin=vmin, vmax=vmax, cmap='viridis', return_scale=True)
            ax.imshow(img)
            plt.axis('off')
            ax.set_title(f'sub {sub} {ses} {label} {title}')
            cbar = plt.colorbar(scale, shrink=0.8, aspect=30)
#             plt.show()
            plt.savefig(f'plots/sub-{sub}_ses-{ses}_{title}_{label}.png')
            plt.clf()
            print(f'plots/sub-{sub}_ses-{ses}_{title}_{label}.png')
```
